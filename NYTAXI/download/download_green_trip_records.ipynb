```python
import pyspark.sql.functions as F

year = 2023
months = [f"{i:02d}" for i in range(2, 13)]

for month in months:
    filename = f"green_tripdata_{year}-{month}.parquet"

    url = f"https://d37ci6vzurychx.cloudfront.net/trip-data/{filename}"
    output_path = f"s3://ims-dbr-landing/nytaxi/trip_records/green/{filename}"

    print(f"Downloading {filename}...")

    df = spark.read.parquet(url)

    df.write \
        .mode("overwrite") \
        .parquet(output_path)

print("Finished downloading all Green 2023 files.")
```
