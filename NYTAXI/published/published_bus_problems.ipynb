#1) Find busiest zones/buroughs for pickup/dropoff and find the total revenue for each zone/burough

##1 borough can contain multiple zones

##(order by trip_count or total_revenue depending on if you want to find busiest zones or total revenue)

###Busiest Pickup Locations by zone & Total revenue by zone.


```python
%sql
select pickup_zone, pickup_borough, count(*) as trip_count, sum(fare_amount) as total_revenue
from nytaxi.published.all_trips 
group by pickup_zone, pickup_borough
order by trip_count desc
```

###Busiest pickup locations by borough & Total revenue by borough.

```python
%sql
select pickup_borough, count(*) as trip_count, sum(fare_amount) as total_revenue
from nytaxi.published.all_trips 
group by pickup_borough
order by trip_count desc
```

###Busiest dropoff locations by zone & Total revenue by zone.

```python
%sql
select dropoff_zone, dropoff_borough, count(*) as trip_count, sum(fare_amount) as total_revenue
from nytaxi.published.all_trips 
group by dropoff_zone, dropoff_borough
order by trip_count desc
```

###Busiest dropoff location by borough & Total revenue by borough.

```python
%sql
select dropoff_borough, count(*) as trip_count, sum(fare_amount) as total_revenue
from nytaxi.published.all_trips 
group by dropoff_borough
order by trip_count desc
```

#2) When do trips happen most(by hour, week, month), and when is the most revenue generated?

###hour (pickup)

```python
%sql
select pu_hour_of_day, count(*) as trip_count, sum(fare_amount) as total_revenue
from nytaxi.published.all_trips 
group by pu_hour_of_day order by trip_count desc;
```

#3)Which taxitype has most total trips? Which taxitype has highest revenue? Avg revenue per taxi trip? Avg trip duration per taxi trip?

###Change the order by depending on what you're trying to find

```python
%sql
select * from nytaxi.published.all_trips
```

```python
%sql
select taxi_type, count(*) as total_trips, sum(fare_amount) as total_revenue, avg(fare_amount) as avg_revenue, avg(trip_duration) as avg_trip_duration 
from nytaxi.published.all_trips
group by taxi_type
order by total_trips desc;
```
