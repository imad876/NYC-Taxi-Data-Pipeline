```python
import pyspark.sql.functions as F
```

```python
source_path = 's3://ims-dbr-landing/nytaxi/zone_lookup/'
output_path = 's3://ims-dbr-published/zone_lookup/'
```

```python
df = spark.read.option('header', True).csv(source_path)
```

```python
df = df.withColumn('LocationID', F.col('LocationID').cast('int'))
display(df)
df.printSchema()
```

```python
## Write to Stage Table
df.write.mode("overwrite").parquet(output_path)
```

```python
%sql
CREATE TABLE IF NOT EXISTS nytaxi.published.zone_lookup
USING PARQUET
LOCATION 's3://ims-dbr-published/zone_lookup/';
```

###all_trips has PULocationID and DOLocationID. zone_lookup has LocationID.

```python
%sql
select * from nytaxi.published.all_trips t 
left join nytaxi.published.zone_lookup pu on t.PULocationID = pu.LocationID 
left join nytaxi.published.zone_lookup do on t.DOLocationID = do.LocationID;
```
