```python
def check_file_exists(path):
    try:
        files = dbutils.fs.ls(path)
        return len(files) > 0
    except Exception as e:
        if 'CloudFileNotFoundException' in str(e) or 'No such file or directory' in str(e):
            return False
        else:
            raise
```
